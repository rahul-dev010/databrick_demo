```python
# Read CSV files from the volume and create a bronze table
catalog = dbutils.widgets.get("catalog")
schema_name = "bronze"
table_name = "accounts"

full_table_name = f"{catalog}.{schema_name}.{table_name}"

input_path = f"/Volumes/{catalog}/bronze/raw/{table_name}/"

# print(f"Reading CSV files from: {input_path}")
# print(f"Target table: {full_table_name}")

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(input_path)
)
# adding the metaa data
df_meta = add_metadata_col(df)

# Write to the bronze table
(
    df_meta.write
    .mode("overwrite")
    .option("overwriteSchema", True)
    .saveAsTable(full_table_name)
)

# print(f"Table {full_table_name} created successfully.")
# display(spark.table(full_table_name).limit(10))
```
