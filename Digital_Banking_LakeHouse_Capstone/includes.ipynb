```python
from pyspark.sql.functions import *

def add_metadata_col(df):
    df1 = (
        df
        .withColumn("file_name", col("_metadata.file_name"))
        .withColumn("file_path", col("_metadata.file_path"))
        .withColumn("ingestion_date", current_timestamp())
    )
    
    return df1
```
